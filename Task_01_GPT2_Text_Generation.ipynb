```python
import torch
from transformers import GPT2Tokenizer, GPT2LMHeadModel, Trainer, TrainingArguments
from datasets import load_dataset
import gradio as gr
```

```python
texts = [
    # Prodigy InfoTech – About
    "Prodigy InfoTech is an e-learning platform that focuses on practical and accessible education.",
    "The mission of Prodigy InfoTech is to transform the way people learn using innovative solutions.",
    "Prodigy InfoTech provides real-world internships to help students gain industry experience.",
    "The company offers internships in web development, machine learning, data science, and generative AI.",
    "Prodigy InfoTech internships focus on hands-on projects and skill development.",

    # Prodigy InfoTech – Internships
    "Generative AI interns at Prodigy InfoTech work on text and image generation tasks.",
    "Internships at Prodigy InfoTech help students apply theoretical knowledge to real-world problems.",
    "The internship programs are designed to prepare students for future careers in technology.",
    "Prodigy InfoTech believes in learning by doing through practical project-based training.",

    # Generative AI Concepts
    "Generative artificial intelligence creates new content based on learned data patterns.",
    "Text generation is one of the most popular applications of generative AI.",
    "Generative models can create text, images, and other forms of content.",
    "Large language models are trained on vast amounts of textual data.",

    # GPT-2 Concepts
    "GPT-2 is a transformer-based language model developed by OpenAI.",
    "GPT-2 generates text by predicting the next word in a sequence.",
    "The transformer architecture uses attention mechanisms to understand context.",
    "GPT-2 can generate coherent and contextually relevant text from a given prompt.",

    # Fine-Tuning Concepts
    "Fine-tuning adapts a pre-trained language model to a specific domain or task.",
    "Fine-tuned models generate text that reflects the style of the training data.",
    "Fine-tuning requires less data compared to training a model from scratch.",
    "Custom datasets help models learn domain-specific language patterns."
]
```

```python
from datasets import Dataset

dataset = Dataset.from_dict({"text": texts})
dataset
from transformers import GPT2Tokenizer, GPT2LMHeadModel

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token

model = GPT2LMHeadModel.from_pretrained("gpt2")
model.resize_token_embeddings(len(tokenizer))
```

```python
def tokenize_function(examples):
    tokens = tokenizer(
        examples["text"],
        truncation=True,
        padding="max_length",
        max_length=64
    )
    # Add labels for Trainer
    tokens["labels"] = tokens["input_ids"]
    return tokens

tokenized_dataset = dataset.map(tokenize_function, batched=True)
tokenized_dataset
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="./results",
    overwrite_output_dir=True,
    num_train_epochs=5,
    per_device_train_batch_size=2,
    save_steps=10,
    save_total_limit=1,
    logging_steps=5,
    report_to="none"
)
```

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset
)

trainer.train()
```

```python
from transformers import GPT2Tokenizer, GPT2LMHeadModel

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

tokenizer.pad_token = tokenizer.eos_token
```

```python
def generate_text(prompt):
    inputs = tokenizer(prompt, return_tensors="pt")
    outputs = model.generate(
        **inputs,
        max_length=50,
        num_return_sequences=1,
        no_repeat_ngram_size=2,
        do_sample=True,
        top_k=50,
        top_p=0.95
    )
    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
import gradio as gr

interface = gr.Interface(
    fn=generate_text,  # function to call
    inputs=gr.Textbox(
        lines=5,
        placeholder="Enter a prompt here...",
        label="Input Prompt"
    ),
    outputs=gr.Textbox(
        lines=8,
        label="Generated Text"
    ),
    title="GPT-2 Text Generator (Prodigy InfoTech)",
    description="Fine-tuned GPT-2 model trained on Prodigy InfoTech + AI concepts. Enter a prompt and see the generated text.",
    theme="default"
)

interface.launch()
```
